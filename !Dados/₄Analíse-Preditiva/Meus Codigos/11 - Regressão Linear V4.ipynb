```python
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

import statsmodels.api as sm
```

```python
['Tempo_de_Carreira'] = df['Tempo_de_Carreira'].str.replace(',','.')
['Tempo_de_Carreira'] = df['Tempo_de_Carreira'].astype('float')
```

```python
df.query("Tempo_de_Carreira < -0")
```

```python
df['Tempo_de_Carreira'] = df['Tempo_de_Carreira'].abs()
```

```python
df['Tempo_de_Carreira'] = df['Tempo_de_Carreira'].astype(str).str[:2]
```

```python
y = df['Salario']
df['interceptor'] = 1
x= df[['Anos_Educ_Superior', 'interceptor', 'Tempo_Empresa', 'Tempo_de_Carreira']]
```

```python
modelo = sm.OLS(y,x)
```

```python
resultado = modelo.fit()
```

```python
print(resultado.summary())
#porcentagem = 75%
#Anos_Edu_Superior = 347.71 - 526.73
#intercepto = 4053.76 - 4301.80
#Tempo_Empresa = 178.45 - 117.66
#Tempo_de_Carreira = -32.46 - 45.1
```
