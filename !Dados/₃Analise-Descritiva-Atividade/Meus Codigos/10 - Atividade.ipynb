## Adicionando Colunas / Atividade

```python
df_final['valor_final'] = df_final['valor_unitário'] * df_final['quantidade']
df_final['valor_final'].head()
```

```python
df_final['valor_custo'] = df_final['valor_custo'].str.replace('R$','')
df_final['valor_custo'] = df_final['valor_custo'].astype('float')
df_final['valor_custo'].head()
```

```python
df_final['lucro_unitário'] = df_final['valor_unitário'] - df_final['valor_custo']
df_final['lucro_unitário']
```

```python
df_final['lucro_registro'] = df_final['quantidade'] * (df_final['valor_unitário'] - df_final['valor_custo'])
df_final['lucro_registro']
```

```python
df_final['lucro_total'] = df_final['quantidade'] * df_final['valor_unitário'] - df_final['valor_custo']
df_final['lucro_total']
```

```python
print(df_final['valor_final'].sum())
```

```python
# Margem de lucro (%) geral:
df_final['margem_lucro'] = (df_final['lucro_total'] / 42555893.0) * 100
```

```python
# Margem de lucro (%) por produto (opcional)
# Adicionando depois da revisão.
df_final['margem_lucro_p'] =(( df_final['valor_unitário'] - df_final['valor_custo']) / df_final['valor_unitário']) * 100
df_final['margem_lucro_p'].head()
```

### criando graficos / atividade

1. Qual foi o faturamento analisado?

```python
print(df_final['valor_final'].sum())
```

2. Qual loja teve o maior faturamento? Grafico com as 5 primeiras lojas de barra.

```python
loja_faturamento = df_final.groupby('id_loja')['valor_final'].sum().sort_values(ascending=False).reset_index()
display(loja_faturamento)
```

```python
fig = px.bar(loja_faturamento, x='id_loja' , y='valor_final', title="loja teve o maior faturamento", color='valor_final', text=loja_faturamento['valor_final'])
fig.show()
```

3. Qual produto foi o mais vendido em quantidade? Grafico com os 3 primeiros barra.

```python
quantidade_vendida = df_final.groupby('produto')['quantidade'].sum().sort_values(ascending=False).reset_index()
display(quantidade_vendida)
```

```python
fig = px.bar(quantidade_vendida, x='produto' , y='quantidade', title="produto foi o mais vendido em quantidade", color='quantidade', text=quantidade_vendida['quantidade'])
fig.show()
```

4. Qual produto foi o menos vendido em quantidade?

Gerente

Qual foi o mês com maior faturamento? Grafico de linha

```python
# comando do professor, não deu tempo de terminar.
df_final['mes'] = df_final['data'].dt.to_period('M')

fat_mes_gerente = df_final.groupby(['mes','gerente'])['valor_final'].sum().reset_index()

fat_mes_gerente['mes'] = fat_mes_gerente['mes'].dt.to_timestamp()

fig = px.line(
    fat_mes_gerente,
    x='mes',
    y='valor_final',
    line_shape='spline',
    markers=True,
    color='gerente'
)

fig.show()
```

Quais foram os produtos mais vendidos por cada gerente? Grafico de arvore

```python
# comando do professor, não deu tempo de terminar.
prod_qtd_gerente = df_final.groupby(['gerente','produto'])['valor_final'].sum().reset_index().sort_values(by='valor_final', ascending=False)

fig = px.treemap(
    prod_qtd_gerente,
    path=[px.Constant('Valor por Gerente'),'gerente','produto'],
    values='valor_final'
)

fig.show()
```
