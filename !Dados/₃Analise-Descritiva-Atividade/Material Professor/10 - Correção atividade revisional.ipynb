##Tratamento dos dados / Data prep

```python
#quando houver caracter especial como $ o regex é necessario.
df_vendas1['Valor Unitário'] = df_vendas1['Valor Unitário'].str.replace(r'R\$', '', regex=True)
```

```python
df_vendas1['Valor Custo'] = df_vendas1['Valor Custo'].str.replace(r'R\$', '', regex=True)
```

```python
#transformação para decimal
df_vendas1 = df_vendas1.astype({
    'Valor Unitário':'float',
    'Valor Custo':float
})
```

```python
df_vendas1['Valor Final'] = df_vendas1['Valor Final'].fillna(df_vendas1['Quantidade'] * df_vendas1['Valor Unitário'])
```

###Unificar as bases

```python
#concatenar as bases de vendas
df_vendas = pd.concat([df_vendas1,df_vendas2], axis=0, join='inner',ignore_index=True)
```

```python
display(df_vendas)
```

```python
#Merge adionar os gerentes
df_final = df_vendas.merge(df_gerentes, how='inner')
```

```python
display(df_final)
```

###Dados duplicados

```python
df_final.duplicated().sum()
```

```python
df_final = df_final.drop_duplicates()
```

```python
df_final['Lucro'] = df_final['Quantidade'] * (df_final['Valor Unitário'] - df_final['Valor Custo'])
df_final['Lucro'].head()
```

```python
#• Margem de lucro (%) por produto (opcional): Margem%_produto = [(Preço Unitário − Custo Unitário) ÷ Preço Unitário] × 100%
df_final['Margem Lucro'] =(( df_final['Valor Unitário'] - df_final['Valor Custo']) / df_final['Valor Unitário']) * 100
df_final['Margem Lucro'].head()
```

##Perguntas de negócio

Qual foi o faturamento analisado?


```python
print(df_final['Valor Final'].sum())
```

Qual loja teve o maior faturamento? Grafico com as 5 primeiras lojas de barra.


```python
fat_loja = df_final.groupby('ID Loja')['Valor Final'].sum().reset_index().sort_values(by='Valor Final', ascending=False).head()

fig = px.bar(
    fat_loja,
    x='ID Loja',
    y='Valor Final',
    text=fat_loja['Valor Final'],
    title='Top 5 Faturamentos Por Loja'
)

fig.update_traces(textposition='outside',texttemplate='R$ %{text:.3s}')

fig.show()
```

Qual produto foi o mais vendido em quantidade? Grafico com os 3 primeiros barra.


```python
prod_qtd = df_final.groupby('Produto')['Quantidade'].sum().reset_index().sort_values(by='Quantidade', ascending=False).head(3)

fig = px.pie(
    prod_qtd,
    names='Produto',
    values='Quantidade',
    title='Top 3 Produtos mais vendidos',
    width= 500, height=500
)

fig.show()
```

Qual produto foi o menos vendido em quantidade?


```python
prod_qtd = df_final.groupby('Produto')['Quantidade'].sum().reset_index().sort_values(by='Quantidade', ascending=False).tail(1)
print(prod_qtd)
```

Qual produto gerou a maior receita?


```python
prod_vf = df_final.groupby('Produto')['Valor Final'].sum().reset_index().sort_values(by='Valor Final', ascending=False).head(1)
print(prod_vf)
```

Qual foi o lucro total e a margem de lucro geral?

```python
lucro_total = df_final['Lucro'].sum()
marg_lucro =  lucro_total / df_final['Valor Final'].sum() * 100
print(f'''
lucro total: {lucro_total:,.2f}
margem lucro geral: {marg_lucro:.2f}
''')
```

Qual foi o mês com maior faturamento? Grafico de linha


```python
df_final['mes'] = df_final['Data'].dt.to_period('M')

fat_mes = df_final.groupby('mes')['Valor Final'].sum().reset_index()

fat_mes['mes'] = fat_mes['mes'].dt.to_timestamp()

fig = px.line(
    fat_mes,
    x='mes',
    y='Valor Final',
    line_shape='spline',
    markers=True
)

fig.show()
```

Qual o faturamento do mes de cada gerente?

```python
df_final['mes'] = df_final['Data'].dt.to_period('M')

fat_mes_gerente = df_final.groupby(['mes','Gerente'])['Valor Final'].sum().reset_index()

fat_mes_gerente['mes'] = fat_mes_gerente['mes'].dt.to_timestamp()

fig = px.line(
    fat_mes_gerente,
    x='mes',
    y='Valor Final',
    line_shape='spline',
    markers=True,
    color='Gerente'
)

fig.show()
```

Quais foram os produtos mais vendidos por cada gerente? Grafico de arvore

```python
prod_qtd_gerente = df_final.groupby(['Gerente','Produto'])['Valor Final'].sum().reset_index().sort_values(by='Valor Final', ascending=False)

fig = px.treemap(
    prod_qtd_gerente,
    path=[px.Constant('Valor por Gerente'),'Gerente','Produto'],
    values='Valor Final'
)

fig.show()
```
