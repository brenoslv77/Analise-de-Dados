```python
#parse_dates - transforma uma ou mais coluna em tipo 'datetime', porem quando o formato for dia-mês-ano é necessario o uso do 'datfirst=True'
df_vendas = pd.read_csv('/content/drive/MyDrive/!dados/datasets/datasets/base_vendas.csv', parse_dates=['dt_venda'], dayfirst=True)
```

```python
df_vendas['dt_entrega'] = pd.to_datetime(df_vendas['dt_entrega'], format='%d/%m/%Y %H:%M')
```

```python
df_vendas.info()
```

```python
df_vendas['preco_custo'].head()
```

```python
nome_colunas = df_vendas.columns
```

```python
#i.replace - trocar espaços por underline
#.lower - texto para minusculo
#.strip - remove espaços estras
#[] - Lista
novos_nomes= []
for i in nome_colunas:
    novos_nomes.append(i.replace(' ','_').lower().strip())
```

```python
print(novos_nomes)
```

```python
df_vendas.columns = novos_nomes
df_vendas.head()
```

```python
#trocando virgula por ponto na coluna
df_vendas['preco_custo'] = df_vendas['preco_custo'].str.replace(',','.')
#transformando a coluna em tipo float
df_vendas['preco_custo'] = df_vendas['preco_custo'].astype('float')
#o programa so funciona uma vez, na segunda tentativa ele mostra "ERROR"
```

```python
#o programa so funciona uma vez, na segunda tentativa ele mostra "ERROR"
df_vendas['valor_unitario'] = df_vendas['valor_unitario'].str.replace(',','.')
df_vendas['valor_unitario'] = df_vendas['valor_unitario'].astype('float')
```

```python
df_vendas['valor_unitario'].head()
```

```python
#apagando as linhas duplicadas
df_vendas = df_vendas.drop_duplicates()
```

##missing values / valores nulo

```python
#apagando linhas vazias
#df_vendas=[df_...].dropna()
```

```python
#preenche manualmente as linhas vazias com valor da sua escolha
df_vendas['comissao'].fillna(0)
```

```python
#preenche com o ultimo valor decima para baixo
df_vendas['comissao'].ffill()
```

```python
#preenche com o ultimo valor de baixo para cima
df_vendas['comissao'].bfill()
```

```python
#todas as comissoes que não estavam preenchida foram colocadas o número citado
df_vendas['comissao'] = df_vendas['comissao'].fillna(0)
```

```python
df_vendas['comissao'] = df_vendas['comissao'].str.replace(',','.')
```

```python
df_vendas = df_vendas.astype({
    'matricula_funcionario':str,
    'codigo_produto':str,
    'comissao':float
})
```

#engenharia de dados

```python
#criando colunas e linhas
df_vendas['valor_final'] = df_vendas['valor_unitario'] * df_vendas['quantidade']
df_vendas['valor_final'].head()
```

```python
df_vendas['lucro'] = (df_vendas['valor_final'] - df_vendas['preco_custo']) * df_vendas['quantidade']
df_vendas['lucro']
```

```python
df_vendas[['descricao_produto','preco_custo','quantidade','valor_final','lucro']]
```

```python
df_vendas['prazo'] = df_vendas['dt_entrega'] - df_vendas['dt_venda']
df_vendas['prazo']
```

```python
df_vendas['margem_lucro'] = df_vendas['lucro'] / df_vendas['valor_final'] * 100
df_vendas['margem_lucro']
```

```python
df_vendas = df_vendas.drop(columns=['matricula_funcionario', 'cargo','codigo_produto','codigo_loja'])
display(df_vendas)
```

```python
#salvando a base pronta
df_vendas.to_csv('/content/drive/MyDrive/!dados/datasets/datasets/arquivo_final.csv', index=None)
```
