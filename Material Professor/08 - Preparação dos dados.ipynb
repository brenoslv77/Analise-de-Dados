```python
import pandas as pd
```

```python
# parse_dates - transforma uma ou mais colunas em tipo 'datetime', porem quando o formato for dia-mes-ano é necessario o uso do 'dayfirst=True'
df_vendas = pd.read_csv('/content/drive/MyDrive/!Dados/base_vendas.csv', parse_dates=['dt_venda'], dayfirst=True)
```

```python
df_vendas['dt_entrega'] = pd.to_datetime(df_vendas['dt_entrega'], format='%d/%m/%Y %H:%M')
```

```python
df_vendas.info()
```

```python
nome_colunas = df_vendas.columns
```

```python
novos_nomes =[]
for i in nome_colunas:
    novos_nomes.append(i.replace(' ','_').lower().strip())
```

```python
print(novos_nomes)
```

```python
df_vendas.columns = novos_nomes
df_vendas.head()
```

```python
#Trocando virgula por ponto na coluna
df_vendas['preco_custo'] = df_vendas['preco_custo'].str.replace(',','.')
#transformando a coluna em tipo float
df_vendas['preco_custo'] = df_vendas['preco_custo'].astype('float')
```

```python
#Trocando virgula por ponto na coluna
df_vendas['valor_unitario'] = df_vendas['valor_unitario'].str.replace(',','.')
#transformando a coluna em tipo float
df_vendas['valor_unitario'] = df_vendas['valor_unitario'].astype('float')
```

```python
#apagando as linhas duplicadas
df_vendas = df_vendas.drop_duplicates()
```

##Missing values / valores nulos

```python
#apagando linhas vazias
df_vendas.dropna()
```

```python
#preenche manualmente as linhas vazias com valor da sua escolha
df_vendas['comissao'].fillna(0.03)
```

```python
#Preencha com o ultimo valor de cima para baixo
df_vendas['comissao'].ffill()
```

```python
#Preencha com o ultimo de baixo para cima
df_vendas['comissao'].bfill()
```

```python
df_vendas['comissao'] = df_vendas['comissao'].fillna(0)
```

```python
df_vendas['comissao'] = df_vendas['comissao'].str.replace(',','.')
```

```python
df_vendas = df_vendas.astype({
    'matricula_funcionario':str,
    'codigo_produto':str,
    'comissao':float
})
```

##Engenharia de dados

```python
df_vendas['valor_final'] = df_vendas['valor_unitario'] * df_vendas['quantidade']
df_vendas['valor_final'].head()
```

```python
df_vendas['lucro'] = df_vendas['valor_final'] - df_vendas['preco_custo'] * df_vendas['quantidade']
df_vendas['lucro']
```

```python
df_vendas['prazo'] =  df_vendas['dt_entrega'] - df_vendas['dt_venda']
df_vendas['prazo']
```

```python
df_vendas['margem_lucro'] = df_vendas['lucro'] / df_vendas['valor_final'] * 100
df_vendas['margem_lucro']
```

```python
df_vendas = df_vendas.drop(columns=['matricula_funcionario','cargo','codigo_loja','codigo_produto'])
display(df_vendas)
```

```python
#Salvando a base pronta
df_vendas.to_csv('/content/drive/MyDrive/!Dados/Arquivo_final.csv', index=None)
```
